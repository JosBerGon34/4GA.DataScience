DEFINICION DEL PROBLEMA:

Queremos averiguar si la posicion polítca de un individuo puede reflejarse en variables socioeconomicas y éticas.

Este estudio solo se realizara en paises concretos de la Union Europea. Se añadira un factor externo a la fuente principal de nuestros datos.

Nos servirá para agrupar los diferentes paises de la Union europea por situaciones económicas similares, este criterio de union para el entreno del modelo se basa principalmente en
 3 grandes conceptos economicos. EL PIB del país, el sueldo anual neto en euros y el porcentaje de inflación.
La diferencia entre el PIB y el SMA nos aclara indirectamente si hay grandes fortunas o hay gran carga fiscal sobre los individuos por lo que subyace un perfil político y cultural a la hora de agrupar.
Si tenemos en cuenta estos dos factores, al incluir la infación de ese país, nos indica la agrupacion(Cluster de país) si el pais en cuestion esta aprovechando bien esos recursos o por el contrario
la mala gestion política esta empobreciendo a la ciudadania y por ende, eso se vera reflejado en los datos del dataset principal.

Al realizar el analisis descriptivo y el EDA, averiguaremos si los criterios económicos externos son diferenciales en los datos, y si son importantes tenerlos en cuenta para predecir la posición política del individuo.
