```python
#Guardamos en un dataframe nuestra tabla para asignar grupo y valores clave de la columna cntry.
import sqlite3
import pandas as pd

# Ruta a tu base de datos SQLite
db_path = 'C:/Users/Josue/4GA.Datascience/4GA.DataScience/src/EcoUE.db'

try:
    # Conectar a la base de datos
    conn = sqlite3.connect(db_path)

    # Consulta SQL para seleccionar todos los datos de la tabla "ecoeu"
    query = "SELECT * FROM ecoeu"

    # Leer los datos en un DataFrame de pandas
    df_ecoeu = pd.read_sql_query(query, conn)

    # Imprimir el DataFrame para verificar
    print(df_ecoeu)

except sqlite3.Error as e:
    print(f"Error al conectar o consultar la base de datos: {e}")

finally:
    # Cerrar la conexión a la base de datos
    if conn:
        conn.close()
```

```python
cntrymap= {
    'ES':'España',
    'IT':'Italia',
    'PT':'Portugal',
    'HR':'Croacia',
}
```

```python
df_media['cntry'] = df_media['cntry'].map(cntrymap)
```

```python
df_media = pd.merge(df_media, df_ecoeu[['cntry', 'cntrycat_factorizado']], on='cntry', how='left')
```
