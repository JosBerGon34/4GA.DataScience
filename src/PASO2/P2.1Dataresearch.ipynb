```python
import sqlite3
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

def scrapping1(url, db_path):

    try:
      
        driver = webdriver.Chrome()  
        driver.get(url)

      
        WebDriverWait(driver, 20).until(
            EC.presence_of_all_elements_located((By.CLASS_NAME, "item"))
        )

      
        items = driver.find_elements(By.CLASS_NAME, "item")
        data = []
        for item in items:
            try:
                country = item.find_element(By.CLASS_NAME, "country-name").text
                year = item.find_elements(By.TAG_NAME, "div")[1].text 
                value = item.find_elements(By.TAG_NAME, "div")[2].text 
                data.append((country, year, value))
            except Exception as e:
                print(f"Error processing item: {e}")

        
        conn = sqlite3.connect(db_path)
        cursor = conn.cursor()

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS inflation_data (
                country TEXT,
                year INTEGER,
                value REAL
            )
        """)

        cursor.executemany("INSERT INTO inflation_data VALUES (?, ?, ?)", data)

        conn.commit()
        conn.close()

        print("Data scraped and stored successfully.")

    except Exception as e:
        print(f"An error occurred: {e}")
    finally:
        try:
            driver.quit()
        except:
            pass


url = "https://datos.bancomundial.org/indicador/FP.CPI.TOTL.ZG?locations=XC"
db_path = "EcoUE.db"  # Path to your SQLite database

scrapping1(url, db_path)


def read_data(db_path):
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM inflation_data")
    rows = cursor.fetchall()
    for row in rows:
        print(row)
    conn.close()

read_data(db_path)
```

```python
import sqlite3
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def scrapping3(url, db_path):
    try:
        service = Service(ChromeDriverManager().install())
        driver = webdriver.Chrome(service=service)
        driver.get(url)

        
        WebDriverWait(driver, 20).until(
            EC.presence_of_element_located((By.ID, "tb1"))
        )

        table = driver.find_element(By.ID, "tb1")
        rows = table.find_elements(By.TAG_NAME, "tr")

        datos = []
        for row in rows[1:]:  
            cells = row.find_elements(By.TAG_NAME, "td")
            if len(cells) >= 6:  
                pais_cell = cells[0].find_element(By.TAG_NAME, "a").get_attribute("title")
                pais = pais_cell.split(" - ")[0]
                salario_eur = cells[5].text.split("€")[0].strip().replace(".", "").replace(",", "") 

                datos.append((pais, int(salario_eur)))

        
        conn = sqlite3.connect(db_path)
        cursor = conn.cursor()
        cursor.execute('''CREATE TABLE IF NOT EXISTS SalarioMedio (pais TEXT, salario_medio_eur INTEGER)''')
        cursor.executemany('INSERT INTO SalarioMedio (pais, salario_medio_eur) VALUES (?, ?)', datos)
        conn.commit()
        conn.close()

        print("Datos extraídos y guardados en la base de datos.")

    except Exception as e:
        print(f"Ocurrió un error: {e}")
    finally:
        try:
            driver.quit()
        except:
            pass


url = "https://datosmacro.expansion.com/mercado-laboral/salario-medio"
db_path = "EcoUE.db"

scrapping3(url, db_path)

def read_data(db_path):
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM SalarioMedio")
    rows = cursor.fetchall()
    for row in rows:
        print(row)
    conn.close()

read_data(db_path)
```

```python
def align_country_names(df_pib, df_inflation):



    # Diccionario de mapeo para nombres de países inconsistentes
    country_mapping = {
        'Unión Europea': 'Zona del Euro',  
        'República Checa': 'Chequia', 
        'República Eslovaca': 'Eslovaquia' 
    }

    # Aplicar el mapeo a ambos DataFrames
    df_pib['country'] = df_pib['country'].replace(country_mapping)
    df_inflation['country'] = df_inflation['country'].replace(country_mapping)

    return df_pib, df_inflation



if df_PIB is not None and df_inflation is not None:
    
    df_PIB, df_inflation = align_country_names(df_PIB.copy(), df_inflation.copy()) 

 
    df_merged = pd.merge(df_PIB, df_inflation, on='country', how='inner') 


    print(df_merged)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.cluster import KMeans
from sklearn.preprocessing import RobustScaler

def grupospaises(df):
    try:
        
        X = df[['PIB', 'Inflation', 'sma']]

        
        scaler = RobustScaler()
        X_scaled = scaler.fit_transform(X)

        
        centroides_iniciales = [
            df[df['cntry'] == 'Alemania'][['PIB', 'Inflation', 'sma']].mean().values, # Rico
            df[df['cntry'] == 'España'][['PIB', 'Inflation', 'sma']].mean().values, # Media Europea
            df[df['cntry'] == 'Hungría'][['PIB', 'Inflation', 'sma']].mean().values # Pobre
        ]

        
        kmeans = KMeans(n_clusters=3, init=scaler.transform(centroides_iniciales), n_init=10, random_state=42)
        df['cntrycat'] = kmeans.fit_predict(X_scaled)

        
        mapeo_categorias = {
            0: 'Rico',
            1: 'Media Europea',
            2: 'Pobre'
        }
        df['cntrycat'] = df['cntrycat'].map(mapeo_categorias)

        
        df['cntrycat_factorizado'] = pd.factorize(df['cntrycat'])[0]

        
        fig = plt.figure(figsize=(12, 8)) 
        ax = fig.add_subplot(111, projection='3d')

        
        ax.scatter(df['PIB'], df['Inflation'], df['sma'], c=df['cntrycat_factorizado'], marker='o', cmap='viridis')

        
        ax.set_xlabel('PIB')
        ax.set_ylabel('Inflación')
        ax.set_zlabel('Salario Medio Anual')

        plt.title('Clusters de Países Europeos (RobustScaler, Factorizado)')
        plt.show()

    except Exception as e:
        print(f"Ocurrió un error: {e}")

grupospaises(df_merged)
```
