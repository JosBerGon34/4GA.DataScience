```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
def visdist(df, columna, ax):
  
    frecuencias = df[columna].value_counts().sort_index()

    sns.barplot(x=frecuencias.index, y=frecuencias.values, color='skyblue', ax=ax)
    ax.plot(frecuencias.index, frecuencias.values, marker='o', color='red')

    ax.set_xlabel('Valor ordinal')
    ax.set_ylabel('Frecuencia')
    ax.set_title(f'Distribución de {columna}')
    ax.set_xticks(frecuencias.index)
    ax.grid(axis='y', linestyle='--')

def vistdistmult(df):
  
    columnas = df.columns
    num_columnas = len(columnas)
    num_filas = (num_columnas + 3) // 4  

    for i in range(num_filas):
        fig, axes = plt.subplots(1, 4, figsize=(26, 8))  

        for j in range(4):
            indice_columna = i * 4 + j
            if indice_columna < num_columnas:
                visdist(df, columnas[indice_columna], axes[j])
            else:
                fig.delaxes(axes[j])  

        plt.tight_layout()
        plt.show()
```

```python
import pandas as pd
import numpy as np

def BinarizePond(df, columnas_personas, columnas_legal, columnas_politica, columnas_cientificos):
    """
    Crea variables binarias de confianza con ajuste probabilístico basado en cuartil y contador de neutralidad.
    """

    # Contador de neutralidad
    columnas_confianza_total = columnas_personas + columnas_legal + columnas_politica + columnas_cientificos
    df['ContNeutro'] = (df[columnas_confianza_total] == 5).sum(axis=1)
    df['ConfNeutro'] = (df['ContNeutro'] >= 5).astype(int)

    # Variables de media binarizadas con ajuste probabilístico basado en cuartil
    df['ConfPplmean'] = df[columnas_personas].mean(axis=1)
    df['ConfLegalMean'] = df[columnas_legal].mean(axis=1)
    df['ConfPoliticaMean'] = df[columnas_politica].mean(axis=1)
    df['ConfSciMean'] = df[columnas_cientificos].mean(axis=1)

    cuartil_personas = df['ConfPplmean'].quantile(0.5)
    cuartil_legal = df['ConfLegalMean'].quantile(0.5)
    cuartil_politica = df['ConfPoliticaMean'].quantile(0.5)
    cuartil_cientificos = df['ConfSciMean'].quantile(0.5)

    def binarizar_con_roll_cuartil(row, media_columna, cuartil):
        if media_columna == 'ConfPplmean':
            columnas_grupo = columnas_personas
        elif media_columna == 'ConfLegalMean':
            columnas_grupo = columnas_legal
        elif media_columna == 'ConfPoliticaMean':
            columnas_grupo = columnas_politica
        else:
            columnas_grupo = columnas_cientificos

        columna_ajuste = columnas_grupo[0]  # Toma la primera columna del grupo

        if row[columna_ajuste] == 5:
            desplazamiento = abs(cuartil - row[media_columna])
            if desplazamiento > 0:
                probabilidad_0 = 0.5 + (desplazamiento / 10)  # Ajusta la probabilidad según el desplazamiento
                if np.random.rand() < probabilidad_0:
                    return 0
                else:
                    return 1
            else:
                return 0  # si el cuartil coincide con 5, asigna 0.
        elif row[media_columna] >= cuartil:
            return 1
        else:
            return 0

    df['ConfPplBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'ConfPplmean', cuartil_personas), axis=1)
    df['ConfLegalBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'ConfLegalMean', cuartil_legal), axis=1)
    df['ConfPoliticaBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'ConfPoliticaMean', cuartil_politica), axis=1)
    df['ConfSciBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'ConfSciMean', cuartil_cientificos), axis=1)

    return df

# Ejemplo de uso:
# Supongamos que tienes un DataFrame llamado 'dfconfi'
# y las columnas de confianza son:
columnas_personas = ['pplfair', 'pplhlp', 'ppltrst']
columnas_politica = ['trstep', 'trstplt', 'trstprl', 'trstprt']
columnas_legal = ['trstlgl', 'trstplc', 'lw_pnd']
columnas_cientificos = ['trtsci_pnd']

BinarizePond(df, columnas_personas, columnas_legal, columnas_politica, columnas_cientificos)

# Ahora dfconfi contiene las nuevas variables binarias
print(df[['ContNeutro', 'ConfNeutro', 'ConfPplBin', 'ConfLegalBin', 'ConfPoliticaBin', 'ConfSciBin']].head())
```

```python
import pandas as pd
import numpy as np

def BinarizePondSat(df, columnas_inmigracion, columnas_gobierno, columnas_dinero):
    """
    Crea variables binarias de satisfacción con ajuste probabilístico basado en cuartil y contador de neutralidad.
    """

    # Contador de neutralidad
    columnas_satisfaccion_total = columnas_inmigracion + columnas_gobierno + columnas_dinero
    df['ContNeutroSat'] = (df[columnas_satisfaccion_total] == 5).sum(axis=1)
    df['SatNeutro'] = (df['ContNeutroSat'] >= 5).astype(int)

    # Variables de media binarizadas con ajuste probabilístico basado en cuartil
    df['SatInmigracionMean'] = df[columnas_inmigracion].mean(axis=1)
    df['SatGobiernoMean'] = df[columnas_gobierno].mean(axis=1)
    df['SatDineroMean'] = df[columnas_dinero].mean(axis=1)

    cuartil_inmigracion = df['SatInmigracionMean'].quantile(0.5)
    cuartil_gobierno = df['SatGobiernoMean'].quantile(0.5)
    cuartil_dinero = df['SatDineroMean'].quantile(0.5)

    def binarizar_con_roll_cuartil(row, media_columna, cuartil):
        if media_columna == 'SatInmigracionMean':
            columnas_grupo = columnas_inmigracion
        elif media_columna == 'SatGobiernoMean':
            columnas_grupo = columnas_gobierno
        else:
            columnas_grupo = columnas_dinero

        columna_ajuste = columnas_grupo[0]  # Toma la primera columna del grupo

        if row[columna_ajuste] == 5:
            desplazamiento = abs(cuartil - row[media_columna])
            if desplazamiento > 0:
                probabilidad_0 = 0.5 + (desplazamiento / 10)  # Ajusta la probabilidad según el desplazamiento
                if np.random.rand() < probabilidad_0:
                    return 0
                else:
                    return 1
            else:
                return 0  # si el cuartil coincide con 5, asigna 0.
        elif row[media_columna] >= cuartil:
            return 1
        else:
            return 0

    df['SatInmigracionBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'SatInmigracionMean', cuartil_inmigracion), axis=1)
    df['SatGobiernoBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'SatGobiernoMean', cuartil_gobierno), axis=1)
    df['SatDineroBin'] = df.apply(lambda row: binarizar_con_roll_cuartil(row, 'SatDineroMean', cuartil_dinero), axis=1)

    return df

# Ejemplo de uso:
# Supongamos que tienes un DataFrame llamado 'dfsat'
# y las columnas de satisfacción son:
col_sat_inm = ['imbgeco', 'imwbcnt']
col_sat_gov = ['stfgov', 'stfdem']
col_sat_money = ['stfeco']

dfsat = BinarizePondSat(df, col_sat_inm, col_sat_gov, col_sat_money)

# Ahora dfsat contiene las nuevas variables binarias
print(df[['ContNeutroSat', 'SatNeutro', 'SatInmigracionBin', 'SatGobiernoBin', 'SatDineroBin']].head())
```

```python
import pandas as pd

def crear_variable_neutro_total(df, columnas_contadores):
    """
    Crea una variable binaria de neutralidad combinada basada en el criterio del 60% de las muestras con valor 5.

    Args:
        df (pd.DataFrame): El DataFrame que contiene los contadores.
        columnas_contadores (list): Una lista con los nombres de las columnas de los contadores.

    Returns:
        pd.DataFrame: El DataFrame con la nueva columna 'NeutroTotal'.
    """

    # 1. Calcular el número total de variables ordinales
    num_variables_ordinales = len(columnas_contadores)

    # 2. Calcular el umbral del 60%
    umbral = num_variables_ordinales * 0.4

    # 3. Contar el número de valores 5 y binarizar
    df['NeutroTotal'] = (df[columnas_contadores] == 5).sum(axis=1) >= umbral
    df['NeutroMean'] = df[['ContNeutroSat', 'ContNeutro']].mean(axis=1)
    # Convertimos la columna NeutroTotal a tipo entero.
    df['NeutroTotal'] = df['NeutroTotal'].astype(int)

    return df

# Ejemplo de uso:
# Supongamos que tienes un DataFrame llamado 'df_fusionado'
# y las columnas de los contadores son:
columnas_contadores = [
    'imbgeco', #1
    'imwbcnt', #2
    'stfgov', #3
    'stfdem', #4
    'stfeco', #5
    'happy',  #6
    'pplfair', #7
    'trstep',  #8
    'trstlgl', #9
    'trtsci_pnd',#10
    'trstplc',  #11
    'lw_pnd',    #12
    'pplhlp',    #13
    'trstplt',  #14
    'trstprl',  #15
    'ppltrst',  #16
    'trstprt',  #17
]  # Reemplaza con tus nombres de columnas reales

df_fusionado = crear_variable_neutro_total(df, columnas_contadores)

# Ahora df_fusionado['NeutroTotal'] contiene la variable de neutralidad combinada binarizada.
df['NeutroTotal'].value_counts()
```

```python
import pandas as pd

def ternariacuartiles(df):

   
    cuartil_25 = df['lrscale'].quantile(0.25)
    cuartil_75 = df['lrscale'].quantile(0.75)

    
    df['lrscale_ternaria'] = 1  

    # Izquierda (0)
    df.loc[df['lrscale'] < (5 - (5 - cuartil_25)), 'lrscale_ternaria'] = 0

    # Derecha (2)
    df.loc[df['lrscale'] > (5 + (cuartil_75 - 5)), 'lrscale_ternaria'] = 2

    return df

df = ternariacuartiles(df)

print(df[['lrscale', 'lrscale_ternaria']].head(10))
print(df['lrscale_ternaria'].value_counts())
```

```python
import pandas as pd

def binarizar_escala_ordinal(df, columna_ordinal):


    # Calcular los percentiles
    percentil_40 = df[columna_ordinal].quantile(0.40)
    percentil_80 = df[columna_ordinal].quantile(0.80)

    # Binarizar la escala ordinal
    df[columna_ordinal + '_binaria'] = 0  # Inicializar con 0

    # Mapear el tercer y cuarto cuartil a 1
    df.loc[df[columna_ordinal] >= percentil_80, columna_ordinal + '_binaria'] = 1

    return df



df = binarizar_escala_ordinal(df, 'polintr')
```
