```python
import pandas as pd
import joblib
import statsmodels.api as sm
from sklearn.model_selection import KFold
from skopt import gp_minimize
from skopt.space import Real, Categorical
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import SelectKBest, f_classif
import os
import numpy as np

def entrenar_y_evaluar_regresion_ordinal(X_train, y_train, X_test, y_test, max_features=None, ruta_guardado=None):
    """
    Entrena y evalúa un modelo de regresión logística ordinal con selección de características y optimización de hiperparámetros.
    """

    original_columns = X_train.columns

    if max_features and max_features < X_train.shape[1]:
        selector = SelectKBest(f_classif, k=max_features)
        X_train = selector.fit_transform(X_train, y_train)
        X_test = selector.transform(X_test)
        selected_features = original_columns[selector.get_support(indices=True)]
    else:
        selected_features = original_columns

    space = {
        'alpha': Real(1e-3, 1.0, prior='log-uniform'),  # Adjusted alpha range
        'method': Categorical(['cloglog', 'logit', 'probit', 'cauchy', 'loglog'])
    }

    def objective(params):
        alpha, method = params

        model = sm.MNLogit(y_train, sm.add_constant(X_train))
        try:
            result = model.fit_regularized(alpha=alpha, method='l1', trim_mode='size', trim_tol=1e-4, method_mi=method, maxiter=20)  # Increased maxiter
            y_pred_probs = result.predict(sm.add_constant(X_train))
            y_pred = np.argmax(y_pred_probs, axis=1)  # Corrected prediction
            accuracy = accuracy_score(y_train, y_pred)
            return -accuracy
        except Exception as e:
            print(f"Error during model fitting: {e}")
            return -0.0  # Return a very low accuracy if fitting fails

    result = gp_minimize(objective, list(space.values()), n_calls=50, random_state=42, verbose=True)

    best_params = {key: value for key, value in zip(space.keys(), result.x)}

    final_model = sm.MNLogit(y_train, sm.add_constant(X_train))
    try:
        final_result = final_model.fit_regularized(alpha=best_params['alpha'], method='l1', trim_mode='size', trim_tol=1e-4, method_mi=best_params['method'], maxiter=20)  # Increased maxiter
        y_pred_probs = final_result.predict(sm.add_constant(X_test))
        y_pred = np.argmax(y_pred_probs, axis=1)  # Corrected prediction
        accuracy = accuracy_score(y_test, y_pred)

        print("\nResultados Finales:")
        print(f"Mejores parámetros: {best_params}")
        print(f"Accuracy en el conjunto de prueba: {accuracy}")
        print(f"Variables utilizadas: {list(selected_features)}")

        if ruta_guardado:
            os.makedirs(os.path.dirname(ruta_guardado), exist_ok=True)
            joblib.dump(final_result, ruta_guardado)
            print(f"Modelo guardado en: {ruta_guardado}")
    except Exception as e:
        print(f"Error during final model fitting: {e}")

# Ejemplo de uso:
X_train, X_test, y1_train, y1_test = joblib.load("C:/Users/Josue/4GA.Datascience/src/PASO5/VariablesTarget/testing/HappyLrscalePolintr/TrTtlrHappyreg.pkl")

entrenar_y_evaluar_regresion_ordinal(X_train, y1_train, X_test, y1_test, max_features=10, ruta_guardado="C:/Users/Josue/4GA.DataScience/models/regresion_ordinal_model.pkl")
```

```python
import numpy as np, random
import pandas as pd
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error
import joblib
from hyperopt import fmin, tpe, hp, STATUS_OK, Trials
from sklearn.model_selection import KFold, cross_val_score
from sklearn.utils import resample
random.seed(42)

def xgboost_regresion_ordinal(X_train, y_train, X_test, y_test, ruta_guardado=None, max_evals=20, max_features=None):
    """
    Función adaptada para regresión ordinal con XGBoost.
    """

    space = {
        'max_depth': hp.quniform('max_depth', 15, 20, 1),
        'learning_rate': hp.uniform('learning_rate', 0.10, 0.13),
        'n_estimators': hp.quniform('n_estimators', 100, 200, 10),
        'subsample': hp.uniform('subsample', 0.85, 0.90),
        'colsample_bytree': hp.uniform('colsample_bytree', 0.42, 0.50),
        'min_child_weight': hp.quniform('min_child_weight', 6, 10, 1),
        'gamma': hp.uniform('gamma', 0.55, 0.65),
        'reg_alpha': hp.loguniform('reg_alpha', np.log(0.2), np.log(0.3)),
        'reg_lambda': hp.loguniform('reg_lambda', np.log(0.05), np.log(0.06))
    }

    def objective(params):
        modelo = xgb.XGBRegressor(
            max_depth=int(params['max_depth']),
            learning_rate=params['learning_rate'],
            n_estimators=int(params['n_estimators']),
            subsample=params['subsample'],
            colsample_bytree=params['colsample_bytree'],
            min_child_weight=int(params['min_child_weight']),
            gamma=params['gamma'],
            random_state=42,
            eval_metric='rmse'  # Changed to rmse
        )
        modelo.fit(X_train, y_train)
        y_pred = modelo.predict(X_test)
        score = mean_squared_error(y_test, y_pred)  # Changed to MSE
        return {'loss': score, 'status': STATUS_OK}  # Minimize MSE

    trials = Trials()
    best = fmin(fn=objective, space=space, algo=tpe.suggest, max_evals=max_evals, trials=trials)

    mejor_modelo = xgb.XGBRegressor(
        max_depth=int(best['max_depth']),
        learning_rate=best['learning_rate'],
        n_estimators=int(best['n_estimators']),
        subsample=best['subsample'],
        colsample_bytree=best['colsample_bytree'],
        min_child_weight=int(best['min_child_weight']),
        gamma=best['gamma'],
        random_state=42,
        eval_metric='rmse'  # Changed to rmse
    )
    mejor_modelo.fit(X_train, y_train, verbose=False)

    if max_features is not None and max_features < X_train.shape[1]:
        feature_importance = mejor_modelo.feature_importances_
        indices_ordenados = np.argsort(feature_importance)[::-1]
        selected_features = X_train.columns[indices_ordenados[:max_features]]
        X_train = X_train[selected_features]
        X_test = X_test[selected_features]
        mejor_modelo.fit(X_train, y_train, verbose=False)

    y_pred = mejor_modelo.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)  # Changed to MSE
    mae = mean_absolute_error(y_test, y_pred) # Added MAE

    kf = KFold(n_splits=10, shuffle=True, random_state=42)
    cv_scores = cross_val_score(mejor_modelo, X_train, y_train, cv=kf, scoring='neg_mean_squared_error') # Changed to neg_mean_squared_error

    print("\nMejor modelo encontrado (XGBoost Regresión Ordinal Hyperopt con Selección de Características):")
    print(f"best_params: {best}")
    print(f"MSE: {mse}")
    print(f"MAE: {mae}")
    print(f"Cross-Validation MSE: {np.mean(cv_scores)}")

    if ruta_guardado:
        joblib.dump(mejor_modelo, f"{ruta_guardado}mejor_modelo_xgboost_regresion_ordinal.pkl")
        with open(f"{ruta_guardado}metricas_xgboost_regresion_ordinal.txt", "w") as f:
            f.write(f"best_params: {best}\n")
            f.write(f"MSE: {mse}\n")
            f.write(f"MAE: {mae}\n")
            f.write(f"Cross-Validation MSE: {np.mean(cv_scores)}\n")
            f.write(f"Predictoras utilizadas: {list(X_train.columns) if isinstance(X_train, pd.DataFrame) else list(range(X_train.shape[1]))}\n")
            f.write(f"Variable objetivo: {y_train.name if hasattr(y_train, 'name') else 'desconocido'}\n")

    return mejor_modelo

# Ejemplo de uso:
X_train, X_test, y1_train, y1_test = joblib.load("C:/Users/Josue/4GA.Datascience/src/PASO5/VariablesTarget/testing/HappyLrscalePolintr/TrTtHappyreg.pkl")
X_train_sampled, y1_train_sampled = resample(X_train, y1_train, n_samples=20000, random_state=42)
Happy3cModel = xgboost_regresion_ordinal(X_train_sampled, y1_train_sampled, X_test, y1_test, ruta_guardado="C:/Users/Josue/4GA.DataScience/models/", max_features=12)
```
