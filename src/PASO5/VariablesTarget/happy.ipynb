```python
#probamos lazypredict
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error
import pandas as pd
import joblib
from hyperopt import hp
# Carga los datos desde el archivo pickle
X_train, X_test, y_train, y_test = joblib.load("C:/Users/Josue/4GA.Datascience/4GA.DataScience/models/Split/happyfloat/splitHappySCL.pkl")


# Definir el espacio de búsqueda de hiperparámetros
espacio_hiperparametros = {
    'alpha': hp.loguniform('alpha', -5, 5),  # Rango logarítmico entre 10^-5 y 10^5
    'l1_ratio': hp.uniform('l1_ratio', 0, 1)  # Rango uniforme entre 0 y 1
}
# Inicializa el modelo ElasticNet
elastic_net = ElasticNet(alpha=1.0, l1_ratio=0.5, random_state=42)

# Entrena el modelo
elastic_net.fit(X_train, y_train)

# Realiza predicciones en el conjunto de prueba
y_pred = elastic_net.predict(X_test)

# Evalúa el rendimiento del modelo
mse = mean_squared_error(y_test, y_pred)
print(f"Error cuadrático medio: {mse}")

# Obtén los coeficientes del modelo
coeficientes = elastic_net.coef_
print(f"Coeficientes: {coeficientes}")
```
